Question 1

```python
from fastembed import TextEmbedding

model = TextEmbedding(model_name="jinaai/jina-embeddings-v2-small-en") # Initialize the embedding model  
```

```python
documents: list[str] = [
    "I just discovered the course. Can I join now?",
]
embeddings = model.embed(documents)  # Get the embeddings for the documents
```

```python
embedding_list = list(embeddings)  # Convert the embeddings to a list
```

```python
len(embedding_list)
```

```python
len(embedding_list[0])  # Check the length of the embedding list
```

```python
min(embedding_list[0])  # Get the minimum value in the embedding list
```

```python
import requests 

docs_url = 'https://github.com/alexeygrigorev/llm-rag-workshop/raw/main/notebooks/documents.json'
docs_response = requests.get(docs_url)
documents_raw = docs_response.json()


documents = []

for course in documents_raw:
    course_name = course['course']
    if course_name != 'machine-learning-zoomcamp':
        continue

    for doc in course['documents']:
        doc['course'] = course_name
        documents.append(doc)
```

```python
q = "I just discovered the course. Can I join now?"  # Query to search for
q_embedding = model_smallest.embed(q)  # Get the embedding for the query
q_embedding_list = list(q_embedding)  # Convert the embedding to a list
```

```python
for doc in documents:
    text = doc['question'] + " " + doc['text']
    embedding_gen = model_smallest.embed([text])  # Get the embeddings for the documents
    embedding_gen_list = list(embedding_gen)  # Convert the embeddings to a list
    doc['embedding'] = text  # Add the full text to the document
    doc['cosine'] = embedding_gen_list[0].dot(q_embedding_list[0])  # Add the embedding to the document
```

```python


list_documents_order = sorted(documents, key=lambda x: x["cosine"], reverse=True)
list_documents_order[0]['cosine']

```
